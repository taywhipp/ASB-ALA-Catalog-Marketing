#### 5.4.1 Overall Body Text Comparison between ALA & ACRL Emails

The side-by-side comparison of TF-IDF terms below reveals distinct content strategies between ALA and ACRL email campaigns:

**ALA Email Body Text** emphasizes promotional language like "school", "management", "librarians", and "public". The emails highlight practical benefits and public accessibility ("webinar", "digital", "accessibility"); and suggest a more general public-oriented approach.

The analysis of high-perofrming ALA eblasts reveals that **engagement is highest when the copy focuses on institutional support and community impact**. Successful emails frequently utilize terms like "community," "resources," and "support," indicating that the ALA audience responds to a mission-driven narrative. The TF-IDF results further show that for ALA, **terms associated with professional "growth" and "management" are highly distinctive in successful campaigns, suggesting that librarians are looking for content that assists in their day-to-day operational success**. While the ACRL audience is driven by "research" and "students", the ALA audience is prioritized around "service" and "infrastructure," making clear, benefit-led language a key driver of their higher open rates (which peaked at nearly 45% in mid-2025).

**ACRL Email Body Text** centers on academic and research-oriented terms like "academic", "research", and "college". The emails typically include professional development and philanthropic keywords ("advocacy", "outreach", "impact"); and indicate a focus on scholarly resources and academic community engagement.

The analysis reveals that high-performing emails (higher open rates) tend to use more specific academic and library-related terms like "book", "library", "new", "research", and "academic" more frequently. Low-performing emails show similar patterns but with less emphasis on these key terms, suggesting that **relevance to library professionals drives engagement**. The TF-IDF analysis highlights that terms like "student" and "academic" are highly distinctive in successful eblasts, indicating that **clear, direct language about library resources resonates better**. Higher performers show stronger TF-IDF scores for terms related to new publications and research materials.

**Strategic Implications** include ALA campaigns prioritizing broad appeal. ACRL campaigns targeting specialized academic audiences with research-focused content. Both organizations effectively use "book" and "library" as core terms, but diverge in secondary messaging. Lastly, current content strategy aligns with target audience expectations and organizational mission.

#### 5.4.4 Key Findings from Email Body Length vs. Engagement

**Body Length vs. Engagement:** Across both ALA and ACRL plots, the data show a weak correlations between email body length and engagement rates. High-performing ALA emails generally have longer bodies (more words) and achieve higher open rates (32.72% vs. 38.10% for low performers), but lower click rates (10.0% vs. 3.2%). High-performing ACRL emails generally have shorter bodies (fewer words) and achieve lower open rates, but higher click rates.

However, it is also worth noting that body text length may also be positively correlated with the number of links within an email, which may positively affect the click rate of longer emails. Additionally, we note that our sample size (10 for ALA; 10 for ACRL) is relatively small, which may limit the scope of the implications.

**Overall Insights:** Successful ALA eblasts are characterized by practical, information-oriented language and broad topical appeal, utilizing terms like "school," "management," and "librarians" to drive engagement.. Successful ACRL individual book promo emails are characterized by professional, library-focused language, detailed subject lines, and shorter body content that provides value to academic audiences.
