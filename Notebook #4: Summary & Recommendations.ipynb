## 1). Spring/Summer 2026 ALA Editions Catalog E-Blast Campaign

| Recommendation | Rationale |
|----------------|-----------|
| Front-load the SS26 launch with a strong first send | SS25 showed launch-week traffic spikes (~1,500 views) that didn't sustain indicating that capturing momentum early is critical. |
| Prioritize school-libraries, infolit, and gifts in early sends | These categories ranked highest in both views and engagement time across SS25 and FW26. |
| Address the click-rate decline by testing tighter, more targeted content | ALA Editions click rates have not recovered since mid-2025. Focused sends around specific titles or categories may drive more intentional click behavior than full catalog blasts. |
|Send ACRL focused emails on Thursday, keep ALA and general focused email sends flexible across M–F | ACRL click rates peak Wednesday–Thursday (~11–12%). ALA Editions does not have a single standout day, making weekday flexibility acceptable.|

## 2). Fall/Winter 2027 Unified ALA Editions + ACRL Catalog



| Recommendation | Rationale |
|----------------|-----------|
| Maintain a dedicated ACRL audience segment even after combining with ALA | The ACRL catalog, particularly the Syllabus Newsletter, consistently outperforms ALA Editions on open and click rates. Merging into a single undifferentiated send risks diluting this engaged audience. |
|Lead subject lines with community- and mission-driven language, avoid promotional terms| Terms like "students," "library workers," and "community college" outperform terms like "free," "ai," and "catalog" by 10–15 percentage points in open rate.|
| Plan email sends that leverage the FW26 increased traffic surge, and specifically the winter "gift" content| FW26 demonstrated that the fall catalog seasons can generate significantly higher and sustained traffic with peaks around 5,700.|
| Use substantive body copy with clear, library-focused language | Higher-performing emails tend to have longer, more specific body content using terms like "book," "library," and "new." Avoid overly brief or generic copy, but don't over-length. |
