## 1.0 ALA Editions

### 1.1 Editions Catalogue SS25

| Variables | Definitions |
|-------------|-------------|
| Path Path & Screen Class | The page path from a website URL and screen class from a mobile app. |
| Views | Reports the total number of times events have been collected. For example, a user who navigates to the 'About Us' page, then the 'Contact us' page, and then back to the 'About Us' page will result in two views of the 'About Us' page and one view of the 'Contact us' page. |
| Active Users | Active users is the primary metric used in Google Analytics to show you how many people have visited your website or used your app. To be considered active, a user needs to have an engaged session or be reported as a new user. See also engaged sessions and new user. |
| Views per User | This metric is calculated by dividing the total number of views by the total number of users. |
| Average Engagement Time per User | Average engagement time tells you the average amount of time people spend on your website or app. This metric is calculated by dividing the total time by the total number of users. To calculate time, a user needs to spend at least one second on a page, and Google Analytics only counts the amount of time when the browser is visible (time is not counted when the browser or tab is hidden or minimized). See also user engagement and average engagement time per session. |
| Event | GA4 collects all data as events. Events will include page views and other automatically collected events using enhanced measurement. |
| Event Count | Shows you the total number of times events have been collected by Google Analytics. |
| Key Event | Key events are used to report on conversions in Google Analytics. Key events should be used to track important actions like purchases, signups, or any other valuable interactions. A key event is reported whenever a user triggers an event that has been defined as a key event. |

- [**GA4 Glossary**](https://www.lovesdata.com/blog/google-analytics-glossary/#event)

### 1.3 Editions e-Blasts

An e-Blast (email blast) is the marketing practice of sending a single, identical message to a large, unsegmented email list simultaneously.

| Variables | Definitions |
|------|---------|
| Name | Internal campaign name/label |
| Subject | Subject line of email |
| Sent | Date sent |

### 1.4 Editions Catalogue Promotions e-Blasts (Compiled from PDFs)

| Variables   | Definitions |
| ----------- | --------------------------------------------------------------------- |
| Date        | The date and time the email was sent |
| Subscribers | The total number of *intended* recipients (emails attempted to send to) |
| Subject     | The subject line shown to recipients |
| **Delivered**   | The number of emails successfully delivered to recipients' inboxes |
| Delivered % | **Delivered** divided by Subscribers (delivery success rate)  |
| Opened      | The number of delivered emails that were opened at least once  |
| Opened %    | Opened divided by **Delivered** (open rate) |
| Clicked     | The number of delivered emails that received at least one click |
| Clicked %   | Clicked divided by **Opened** (click-through rate) |
| Unsubs      | The number of recipients who unsubscribed after receiving the email |
| Unsubs %    | Unsubs divided by **Delivered** (unsubscribe rate) |
